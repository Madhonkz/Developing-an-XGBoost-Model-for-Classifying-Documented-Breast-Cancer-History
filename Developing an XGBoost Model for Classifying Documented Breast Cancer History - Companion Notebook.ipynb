## 1. Data preparation

This cell loads the three BCSC Risk Factors Dataset CSV files, filters to the 2013–2017 study window, and prepares the data for model training. 
Rows where the outcome variable (`breast_cancer_history`) is coded as 9 (unknown) are removed, as these cannot be used for supervised learning. 
The sentinel code 9 (used across multiple predictor columns to indicate unknown values) is recoded to `NaN` to allow XGBoost's built-in sparsity-aware split-finding mechanism to handle missing values natively without imputation. The two nominal (unordered) predictors, `race_eth` and `menopaus` , are cast to pandas `category` dtype before recoding, as performing the replacement first would force a floating-point conversion that XGBoost's categorical handling does not support.

````python
# Load the dataset's .csv files
CSV_FILES = [
    "bcsc_risk_factors_summarized1_092020.csv",
    "bcsc_risk_factors_summarized2_092020.csv",
    "bcsc_risk_factors_summarized3_092020.csv",
]
# Use only rows where the year is between 2013 and 2017
YEAR_MIN, YEAR_MAX = 2013, 2017

# Per the BCSC Risk Factors Dataset Documentation: code 9 = "Unknown" for these columns.
# age_group_5_years has no unknown code (always observed at the index exam).
UNKNOWN_9_COLS = ["race_eth", "first_degree_hx", "age_menarche", "age_first_birth",
                   "BIRADS_breast_density", "current_hrt", "menopaus", "bmi_group",
                   "biophx", "breast_cancer_history"]
NOMINAL_COLS = ["race_eth", "menopaus"]

#Checks that all three CSVs are present in the folder. Otherwise, shows an error message.
missing = [p for p in CSV_FILES if not Path(p).exists()]
if missing:
    raise FileNotFoundError(f"Could not find: {missing}. Place the 3 BCSC CSVs next to this notebook.")

# Reads all three CSVs and stacks them into one single table.
df = pd.concat([pd.read_csv(p) for p in CSV_FILES], ignore_index=True)
df = df[(df.year >= YEAR_MIN) & (df.year <= YEAR_MAX)].copy()

# Records the total number of weighted screening exams before the next filtering step, for documentation purposes.
n_weighted_before = int(df["count"].sum())
# Removes rows where the outcome is unknown
df = df[df.breast_cancer_history != 9].copy()
#Records the weighted total after that removal, to show how many exams were removed due to unknown outcomes.
n_weighted_after = int(df["count"].sum())

# Nominal (unordered) columns: cast to category FIRST (categories stay plain int64),
# then drop the "9 = Unknown" category so it becomes a proper missing value. Doing
# replace(9, NaN) before astype("category") would force the categories to float64,
# which XGBoost's categorical split handling rejects outright.
for c in NOMINAL_COLS:
    df[c] = df[c].astype("category")
    df[c] = df[c].cat.remove_categories([9])

# Ordinal/numeric columns: plain NaN recode is fine, since these stay numeric.
for c in [c for c in UNKNOWN_9_COLS if c != "breast_cancer_history" and c not in NOMINAL_COLS]:
    df[c] = df[c].replace(9, np.nan)

# After all the filtering, row numbers may have gaps. This resets them to a clean 0, 1, 2, 3... sequence.
df = df.reset_index(drop=True)

# Prints original population size, population after removing entries with unknown outcomes, number of unique rows, number of events, and that the category dtypes are correct integers.
print(f"Weighted N, 2013-2017, before outcome filter: {n_weighted_before:,}")
print(f"Weighted N, after dropping unknown-outcome rows: {n_weighted_after:,}")
print(f"Rows (unique covariate patterns): {len(df):,}")
print(f"Weighted events (breast_cancer_history=1): {int(df.loc[df.breast_cancer_history==1,'count'].sum()):,}")
print(f"race_eth categories: {df['race_eth'].cat.categories.tolist()} (dtype: {df['race_eth'].cat.categories.dtype})")
print(f"menopaus categories: {df['menopaus'].cat.categories.tolist()} (dtype: {df['menopaus'].cat.categories.dtype})")
````

````python
# Missingness per predictor, weighted by count
total_w = df["count"].sum()
for c in [c for c in UNKNOWN_9_COLS if c != "breast_cancer_history"]:
    n_unknown = df.loc[df[c].isna(), "count"].sum()
    print(f"{c:25s} {n_unknown:>10,.0f}  ({n_unknown/total_w*100:5.2f}%)")
````

## 2. Train XGBoost model and evaluate apparent performance

### 2.1. Training the XGBoost model

This cell defines the feature set, target variable, and sample weights, then trains the XGBoost model on the full 2013–2017 dataset. No train/test split is used, as the entire subset is used for development, with internal validation handled separately via bootstrap resampling in Section 3. Each row is weighted by its `count` value, so the model learns from the true population distribution rather than treating each unique covariate combination equally. Class imbalance is addressed through `scale_pos_weight`, calculated as the ratio of weighted negatives to weighted positives (9.354), which up-weights the minority class during training. The trained model is saved to `xgb_bcsc_model.json` for reproducibility, and apparent predicted probabilities are generated on the full dataset for use in the evaluation cells below.

````python
# Defines which columns from the CSVs are the predictors, the outcome and the population counts
FEATURES = ["age_group_5_years", "race_eth", "first_degree_hx", "age_menarche",
            "age_first_birth", "BIRADS_breast_density", "current_hrt", "menopaus",
            "bmi_group", "biophx"]
TARGET = "breast_cancer_history"
WEIGHT = "count"

# Splits the table in three parts: input features, outcome labels (converted to integers 0/1), weights (converted to decimals)
X = df[FEATURES]
y = df[TARGET].astype(int)
w = df[WEIGHT].astype(float)

# Calculates the weighted number of exams without (neg) and with (pos) documented breast cancer history
neg, pos = w[y == 0].sum(), w[y == 1].sum()
scale_pos_weight = neg / pos
print(f"Weighted negatives: {neg:,.0f}  Weighted positives: {pos:,.0f}  scale_pos_weight={scale_pos_weight:.3f}")

# Sets the model's hyperparameters
XGB_PARAMS = dict(
    objective="binary:logistic", eval_metric="auc", max_depth=4, learning_rate=0.05,
    n_estimators=300, subsample=0.8, colsample_bytree=0.8, reg_lambda=1.0,
    enable_categorical=True, random_state=42, n_jobs=-1,
)

# Trains the model, weighting each covariate pattern (row) by the number of exams it represents. A row with count=500 affects the model's learning 500 times more than a row with count=1.
model = xgb.XGBClassifier(**XGB_PARAMS, scale_pos_weight=scale_pos_weight)
model.fit(X, y, sample_weight=w)

# Saves the trained model to a file so it can be reloaded later without retraining.
model.save_model("xgb_bcsc_model.json")

# Runs the trained model on the same data it was trained on to get predicted probabilities.
p_apparent = model.predict_proba(X)[:, 1]
print("Model trained and saved to xgb_bcsc_model.json")
````

### 2.2 Apparent performance metrics

Before running bootstrap validation, we compute the model's *apparent* performance, its performance on the same data it was trained on. These figures are deliberately optimistic and will later be corrected for overfitting in Section 3.

Two metrics are computed:

- **AUC-ROC** (`weighted_auc`): Discrimination - the probability that the model ranks a positive case higher than a negative one. Weighted by `count` to reflect the number of screening exams each row represents.

- **Calibration slope and intercept** (`calibration_slope_intercept`): Calibration - how well predicted probabilities match the observed proportions of exams with documented history. Computed via logistic (Cox) calibration: an unpenalized logistic regression of true outcomes on the logit  of predicted probabilities. A perfectly calibrated model has slope = 1.0 and  intercept = 0.0. The large negative intercept here is an expected artifact of `scale_pos_weight` and is addressed in Section 2.6.

````python
# Adds the `count` column as weight for the AUC calculations to keep it consistent with the model's training
def weighted_auc(y_true, y_score, weights):
    return roc_auc_score(y_true, y_score, sample_weight=weights)

# Computes logistic (Cox) calibration: the slope and intercept of a weighted logistic regression of the outcome on the logit of the predictions. 
def calibration_slope_intercept(y_true, y_score, weights):
    """Fit y ~ logit(p_hat) via weighted logistic regression (flexible/Cox calibration)."""
    eps = 1e-6
    p = np.clip(y_score, eps, 1 - eps)
    logit_p = np.log(p / (1 - p))
    lr = LogisticRegression(C=1e10, solver="lbfgs", max_iter=200)  # ~unpenalized
    lr.fit(logit_p.reshape(-1, 1), y_true, sample_weight=weights)
    return float(lr.coef_[0][0]), float(lr.intercept_[0]), logit_p

# Apparent predictions on the same data that the model was trained on. This is intentionally optimistic and will be corrected through Bootstrap Internal Validation.
auc_apparent = weighted_auc(y, p_apparent, w)
slope_apparent, intercept_apparent, logit_p = calibration_slope_intercept(y.values, p_apparent, w.values)

# Prints the results
print(f"Apparent AUC-ROC: {auc_apparent:.4f}")
print(f"Apparent calibration slope: {slope_apparent:.4f}, intercept: {intercept_apparent:.4f}")
````

### 2.3 ROC curve

The ROC curve plots sensitivity (true positive rate) against 1 - specificity (false positive rate) across all classification thresholds, weighted by `count`. The diagonal dashed line represents a random classifier (AUC = 0.500); the further the curve bows toward the top-left corner, the stronger the model's discrimination. The AUC shown is the apparent value computed in Section 2.2. The figure is saved to `fig_roc_curve.png`.

````python
# Records all true positives and false positives across every possible classification and adds the same weight (w) as all other sections to the ROC curve.
fpr, tpr, _ = roc_curve(y, p_apparent, sample_weight=w)

# Designs the figure
plt.figure(figsize=(5.5, 5.5))
plt.plot(fpr, tpr, color="#1f5fa8", lw=2, label=f"XGBoost (AUC = {auc_apparent:.3f})")
plt.plot([0, 1], [0, 1], color="grey", lw=1, linestyle="--", label="Chance (AUC = 0.500)")
plt.xlabel("1 - Specificity (False Positive Rate)")
plt.ylabel("Sensitivity (True Positive Rate)")
plt.title("AUC-ROC Curve: XGBoost BCSC Model")
plt.legend(loc="lower right")
plt.tight_layout()

# Saves the figure in a PNG and shows the figure within the codebook.
plt.savefig("fig_roc_curve.png", dpi=200)
plt.show()
````

### 2.4 Hyperparameter check: cross-validated random search

The hyperparameters in Section 2.1 were originally set manually. This section checks whether a systematic, cross-validated search would have selected a clearly better configuration.

**Design**
- **Search method:** random search over a pre-specified space (Bergstra & Bengio, 2012). With 60 randomly sampled configurations, the probability that at least one lies in the best-performing 5% of the search space is 1 − 0.95⁶⁰ ≈ 0.95.
- **Search space:** `max_depth` 3–8; `learning_rate` 0.02–0.3 (log-uniform); `min_child_weight` 1–100 (log-uniform); `subsample` 0.6–1.0; `colsample_bytree` 0.6–1.0; `reg_lambda` 0.1–10 (log-uniform). Class weighting (`scale_pos_weight`) and count weights are kept as in the main model.
- **Number of trees:** chosen by early stopping (maximum 2,000 trees; stop after 50 rounds without improvement), using a separate 10% inner split of each training fold. The fold used to score a configuration is never used to choose its number of trees, so the CV estimate is not optimistically biased.
- **Cross-validation:** stratified 5-fold CV over rows (unique covariate patterns), so all exams sharing a pattern fall in the same fold. All models are trained on the same 90% inner-training portion of each fold and scored with the count-weighted AUC used throughout the notebook.
- **Reference:** the current configuration (Section 2.1, exactly 300 trees, no early stopping) is evaluated under identical conditions.

**Pre-specified decision rule** (set before running): the current configuration is **retained** if the best configuration's mean CV AUC exceeds the current configuration's by **less than 0.002**, a margin larger than the bootstrap optimism of the main model and smaller than any practically meaningful difference. Otherwise, the best configuration is **adopted**: update `XGB_PARAMS` in Section 2.1 with the printed values and re-run the notebook from the top.

**Runtime:** 61 configurations × 5 folds. Results are saved after every fit to `tuning_cv_results.csv`; re-running the cell resumes where it stopped.

````python
from sklearn.model_selection import StratifiedKFold, train_test_split

# Search settings: number of random configurations, folds, early-stopping limits and the
# pre-specified decision margin. Changing DECISION_MARGIN after seeing the results would invalidate the check.
N_CONFIGS = 60          # random configurations (plus the current one)
N_FOLDS = 5
MAX_TREES, PATIENCE = 2000, 50
DECISION_MARGIN = 0.002  # pre-specified; do not change after seeing results
tune_path = Path("tuning_cv_results.csv")

# Samples all configurations up front with a fixed seed, so the search is reproducible
rng_tune = np.random.default_rng(2026)
# Draws a value whose logarithm is uniform between lo and hi, so that small and large values
# (e.g. learning rates of 0.02 and 0.2) are sampled equally often on a relative scale.
def loguniform(lo, hi):
    return float(np.exp(rng_tune.uniform(np.log(lo), np.log(hi))))

# Configuration 0 is the manual configuration from Section 2.1 (fixed 300 trees, no early stopping),
# included as the reference against which the random configurations are compared.
configs = [{"config_id": 0, "label": "current (Section 2.1)",
            "max_depth": XGB_PARAMS["max_depth"], "learning_rate": XGB_PARAMS["learning_rate"],
            "min_child_weight": XGB_PARAMS.get("min_child_weight", 1.0),
            "subsample": XGB_PARAMS["subsample"], "colsample_bytree": XGB_PARAMS["colsample_bytree"],
            "reg_lambda": XGB_PARAMS["reg_lambda"], "early_stopping": False}]
# Configurations 1–60 are drawn at random from the pre-specified search space.
for k in range(1, N_CONFIGS + 1):
    configs.append({"config_id": k, "label": f"random {k}",
                    "max_depth": int(rng_tune.integers(3, 9)),
                    "learning_rate": loguniform(0.02, 0.3),
                    "min_child_weight": loguniform(1, 100),
                    "subsample": float(rng_tune.uniform(0.6, 1.0)),
                    "colsample_bytree": float(rng_tune.uniform(0.6, 1.0)),
                    "reg_lambda": loguniform(0.1, 10),
                    "early_stopping": True})
configs_df = pd.DataFrame(configs)
configs_df.to_csv("tuning_configs.csv", index=False)

# Folds are fixed once, so every configuration is evaluated on identical splits
# Stratified 5-fold split over rows: each fold keeps the same share of history-positive rows.
# Within each training fold, 10% is set aside only for early stopping (choosing the number of trees),
# so the fold used for scoring is never used for any choice.
skf = StratifiedKFold(n_splits=N_FOLDS, shuffle=True, random_state=42)
folds = []
for f, (tr_idx, te_idx) in enumerate(skf.split(X, y)):
    fit_idx, es_idx = train_test_split(tr_idx, test_size=0.10, stratify=y.iloc[tr_idx], random_state=f)
    folds.append((fit_idx, es_idx, te_idx))

# Resume logic: (configuration, fold) pairs already saved in the CSV are skipped.
done = set()
if tune_path.exists():
    prev = pd.read_csv(tune_path)
    done = set(zip(prev["config_id"], prev["fold"]))
    print(f"Resuming: {len(done)} of {len(configs) * N_FOLDS} fits already completed.")

# Settings shared by all configurations (objective, class weighting, categorical handling, seed);
# the searched hyperparameters are removed here and supplied per configuration.
base = {k: v for k, v in XGB_PARAMS.items()
        if k not in ("max_depth", "learning_rate", "subsample", "colsample_bytree",
                     "reg_lambda", "min_child_weight", "n_estimators")}

# Main loop: fit every configuration on every fold and record its weighted AUC on the held-out fold.
for cfg in configs:
    for f, (fit_idx, es_idx, te_idx) in enumerate(folds):
        if (cfg["config_id"], f) in done:
            continue
        hp = {k: cfg[k] for k in ("max_depth", "learning_rate", "min_child_weight",
                                  "subsample", "colsample_bytree", "reg_lambda")}
        # Random configurations: up to MAX_TREES trees, stopping once the AUC on the inner 10% split
        # has not improved for PATIENCE rounds. The manual configuration keeps its fixed 300 trees.
        if cfg["early_stopping"]:
            m = xgb.XGBClassifier(**base, **hp, n_estimators=MAX_TREES,
                                  early_stopping_rounds=PATIENCE, scale_pos_weight=scale_pos_weight)
            m.fit(X.iloc[fit_idx], y.iloc[fit_idx], sample_weight=w.iloc[fit_idx],
                  eval_set=[(X.iloc[es_idx], y.iloc[es_idx])],
                  sample_weight_eval_set=[w.iloc[es_idx]], verbose=False)
            n_trees = m.best_iteration + 1
        else:
            n_trees = XGB_PARAMS["n_estimators"]
            m = xgb.XGBClassifier(**base, **hp, n_estimators=n_trees, scale_pos_weight=scale_pos_weight)
            m.fit(X.iloc[fit_idx], y.iloc[fit_idx], sample_weight=w.iloc[fit_idx])
        # Predicts the held-out fold with the selected number of trees and saves the weighted AUC immediately.
        p_te = m.predict_proba(X.iloc[te_idx], iteration_range=(0, n_trees))[:, 1]
        row = {"config_id": cfg["config_id"], "fold": f, "n_trees": n_trees,
               "cv_auc": weighted_auc(y.iloc[te_idx], p_te, w.iloc[te_idx])}
        pd.DataFrame([row]).to_csv(tune_path, mode="a", header=not tune_path.exists(), index=False)
    print(f"  config {cfg['config_id']:>2}/{N_CONFIGS} done")

print("Random search complete.")
````

#### 2.4.1 Results and decision

Summarises the search: mean and standard deviation (SD) of the cross-validated AUC per configuration across the five folds, the median number of trees chosen by early stopping, and the rank of the manual configuration. The pre-specified decision rule is then applied automatically. In the thesis run, the best configuration improved on the manual one by 0.0013 AUC, below the 0.002 margin, so the manual configuration was retained (thesis Section 4.4).

````python
# Loads the saved search results and keeps only configurations evaluated on all five folds.
res = pd.read_csv(tune_path)
counts = res.groupby("config_id").size()
complete_ids = counts[counts == N_FOLDS].index
# Mean and SD of the CV AUC per configuration, the median number of trees, and the sampled hyperparameters.
summary = (res[res["config_id"].isin(complete_ids)]
           .groupby("config_id")
           .agg(mean_cv_auc=("cv_auc", "mean"), sd_cv_auc=("cv_auc", "std"),
                median_trees=("n_trees", "median"))
           .join(configs_df.set_index("config_id"))
           .sort_values("mean_cv_auc", ascending=False))

# Compares the best configuration with the manual configuration (id 0).
current = summary.loc[0]
best_id = summary.index[0]
best = summary.loc[best_id]
gain = best["mean_cv_auc"] - current["mean_cv_auc"]
current_rank = int(np.where(summary.index == 0)[0][0]) + 1

cols = ["label", "mean_cv_auc", "sd_cv_auc", "median_trees", "max_depth", "learning_rate",
        "min_child_weight", "subsample", "colsample_bytree", "reg_lambda"]
summary[cols].round(4).to_csv("tuning_summary.csv")
print(f"Configurations evaluated: {len(summary)} (of {len(configs)})")
print("Top 10 configurations by mean CV AUC:")
display(summary[cols].head(10).round(4))
print("Current configuration:")
display(summary.loc[[0], cols].round(4))

print(f"\nCurrent configuration mean CV AUC: {current['mean_cv_auc']:.4f} (SD {current['sd_cv_auc']:.4f}), "
      f"rank {current_rank} of {len(summary)}")
print(f"Best configuration (id {best_id}) mean CV AUC: {best['mean_cv_auc']:.4f} (SD {best['sd_cv_auc']:.4f})")
print(f"Difference (best − current): {gain:+.4f}  |  pre-specified margin: {DECISION_MARGIN}")

# Applies the pre-specified decision rule.
if gain < DECISION_MARGIN:
    print("\nDECISION: RETAIN the current configuration (difference below the pre-specified margin).")
else:
    new_params = {k: (int(best[k]) if k == "max_depth" else round(float(best[k]), 4))
                  for k in ("max_depth", "learning_rate", "min_child_weight",
                            "subsample", "colsample_bytree", "reg_lambda")}
    new_params["n_estimators"] = int(round(best["median_trees"]))
    print("\nDECISION: ADOPT the best configuration. Update XGB_PARAMS in Section 2.1 with:")
    print(new_params)
    print("then re-run the notebook from the top. Delete bootstrap result CSVs from earlier runs first.")
````

### 2.5 Calibration plot (raw)

Calibration is assessed by grouping predictions into ten deciles of predicted probability and comparing the mean predicted probability in each decile with the observed proportion of exams with documented history, both weighted by `count`. Points on the diagonal indicate perfect calibration. The model is expected to over-predict, because `scale_pos_weight` shifts its predictions upwards during training. The calibration *slope* (≈ 1.04) from Section 2.2 shows that the spread of predictions is appropriate; only the intercept is offset, and this is corrected in Section 2.6.
The figure is saved to `fig_calibration_plot.png`.

````python
#Groups predictions in 10 deciles by predicted probability and, for each decile, calculates the average predicted probability and the average observed outcome rate.
def decile_calibration(pred, y_true, weights):
    tmp = pd.DataFrame({"p": pred, "y": y_true, "w": weights})
    tmp["decile"] = pd.qcut(tmp["p"], 10, labels=False, duplicates="drop")
    return tmp.groupby("decile").apply(lambda d: pd.Series({
        "mean_pred": np.average(d["p"], weights=d["w"]),
        "obs_rate": np.average(d["y"], weights=d["w"]),
    })).reset_index()

# Runs the function on the raw, uncorrected predictions
cal = decile_calibration(p_apparent, y.values, w.values)

# Designs the figure
plt.figure(figsize=(5.5, 5.5))
plt.plot([0, cal["mean_pred"].max() * 1.05], [0, cal["mean_pred"].max() * 1.05],
          color="grey", lw=1, linestyle="--", label="Perfect calibration")
plt.plot(cal["mean_pred"], cal["obs_rate"], marker="o", color="#c0392b", lw=2,
          label="XGBoost (deciles of predicted risk)")
plt.xlabel("Mean predicted probability")
plt.ylabel("Observed proportion with prior breast cancer")
plt.title("Calibration Plot: XGBoost BCSC Model (raw)")
plt.legend(loc="upper left")
plt.tight_layout()

# Saves the figure in a PNG and shows the figure within the codebook.
plt.savefig("fig_calibration_plot.png", dpi=200)
plt.show()
cal
````

### 2.6 Calibration plot (recalibrated)

Logistic recalibration rescales the model's log-odds predictions using the calibration slope and intercept estimated in Section 2.2, then converts them back to probabilities with the sigmoid function. This corrects the intercept offset introduced by `scale_pos_weight`, re-anchoring predicted probabilities to the observed prevalence (~9.66%). Because the transformation preserves the order of predictions, discrimination (AUC) is unchanged. The decile calibration plot is then repeated on the recalibrated predictions.

Note that the recalibration is estimated on the same data on which it is plotted, so close agreement is expected by construction; calibration beyond the development data is assessed by the optimism-corrected calibration slope in Section 3.2.
The figure is saved to `fig_calibration_plot_recalibrated.png`.

````python
# Rescales the logit from the 2.2 predictions using the calibration slope and intercept, and converts it back to probability.
p_recal = 1 / (1 + np.exp(-(slope_apparent * logit_p + intercept_apparent)))
cal_recal = decile_calibration(p_recal, y.values, w.values)

# Designs the figure
plt.figure(figsize=(5.5, 5.5))
plt.plot([0, 1], [0, 1], color="grey", lw=1, linestyle="--", label="Perfect calibration")
plt.plot(cal_recal["mean_pred"], cal_recal["obs_rate"], marker="o", color="#1f8a4c", lw=2,
          label="XGBoost (recalibrated)")
plt.xlabel("Mean predicted probability (recalibrated)")
plt.ylabel("Observed proportion with prior breast cancer")
plt.title("Calibration Plot After Logistic Recalibration")
plt.legend(loc="upper left")
plt.tight_layout()

# Saves the figure in a PNG and shows the figure within the codebook. Displays `mean_pred` and `obs_rate`.`
plt.savefig("fig_calibration_plot_recalibrated.png", dpi=200)
plt.show()
cal_recal
````

## 3. Model Validation

### 3.1 Bootstrap Internal Validation

Optimism-corrected bootstrap validation following Harrell's method. For each of B = 200 bootstrap iterations, a new XGBoost model is trained on a bootstrap resample (rows, i.e. unique covariate patterns, drawn with replacement from the full dataset) and evaluated twice: once on the bootstrap sample itself (`auc_boot_train`) and once on the original full dataset (`auc_boot_test`). The difference between the two is the *optimism* for that iteration: how much the model overestimates its performance when tested on its own training data.

Mean optimism across all 200 iterations is subtracted from the apparent AUC and calibration slope to produce optimism-corrected estimates. A 95% confidence interval is derived from the 2.5th and 97.5th percentiles of the per-iteration optimism-corrected AUCs. Results are saved incrementally to `bootstrap_results.csv`; re-running this cell after an interruption resumes from the last completed iteration.

Note: the bootstrap treats rows (and the exams they represent) as independent. As a woman can contribute up to one exam per calendar year and the public data contain no individual identifier, the confidence interval is likely somewhat too narrow. An alternative bootstrap resampling individual exams is used in Section 5.

````python
# Number of bootstrap iterations. Reduce for a faster/rougher estimate
B = 200  

# Checkpoint/resume logic. Results are saved incrementally, so if the kernel crashes mid-run, it picks up from where it left off rather than starting over.
out_path = Path("bootstrap_results.csv")
done_iters = set()
if out_path.exists():
    done_iters = set(pd.read_csv(out_path)["iter"].tolist())
    print(f"Resuming: {len(done_iters)} iterations already completed.")

# Total number of rows in the data set. `rows` is an empty list that will collect results from each iteration.
n = len(df)
rows = []

# Loop from 0 to 199. If this iteration number was already saved in the CSV from a previous run, skip it.
for b in range(B):
    if b in done_iters:
        continue
    # Creates a random number generator so each bootstrap is reproducible. 
    rng = np.random.default_rng(1000 + b)
    # Allows the bootstrap to draw random rows with replacement for the validation.
    boot_idx = rng.integers(0, n, size=n)
    # Uses indices to slice out the bootstrap sample.    
    Xb, yb, wb = X.iloc[boot_idx], y.iloc[boot_idx], w.iloc[boot_idx]
    # Trains a brand new model on the bootstrap sample using the exact same hyperparameters as the main model.
    mb = xgb.XGBClassifier(**XGB_PARAMS, scale_pos_weight=scale_pos_weight)
    mb.fit(Xb, yb, sample_weight=wb)
    # Generates two prediction sets from the same bootstrap model. The trained, optimistic one and the new, realistic one.
    p_bb = mb.predict_proba(Xb)[:, 1]
    p_bo = mb.predict_proba(X)[:, 1]
    # Computes AUC and calibration slope for both sets.
    auc_bt = weighted_auc(yb, p_bb, wb)
    auc_bo = weighted_auc(y, p_bo, w)
    s_bt, _, _ = calibration_slope_intercept(yb.values, p_bb, wb.values)
    s_bo, _, _ = calibration_slope_intercept(y.values, p_bo, w.values)
    # Saves the four numbers for this iteration into the `rows` list as a dictionary.
    rows.append({"iter": b, "auc_boot_train": auc_bt, "auc_boot_test": auc_bo,
                  "slope_boot_train": s_bt, "slope_boot_test": s_bo})
    # Every 10 iterations prints a progress update, flushes the accumulated rows to the CSV file, then clear rows to free memory.
    if (b + 1) % 10 == 0:
        print(f"  iter {b}: auc_test={auc_bo:.4f}  slope_test={s_bo:.4f}")
        pd.DataFrame(rows).to_csv(out_path, mode="a", header=not out_path.exists(), index=False)
        rows = []
        
# After the loop ends, writes any remaining rows that didn't land on a multiple of 10.
if rows:
    pd.DataFrame(rows).to_csv(out_path, mode="a", header=not out_path.exists(), index=False)

print("Bootstrap complete.")
````

### 3.2 Results Summary

Reads the 200 iterations saved to `bootstrap_results.csv` and computes the final validated performance estimates. For each iteration, the difference between boot-train and boot-test AUC represents the optimism introduced by testing a model on its own training data. The mean optimism across all 200 iterations (0.0012) is subtracted from the apparent AUC to produce the optimism-corrected AUC (0.8956), with a 95% confidence interval of 0.8938–0.8971 from the percentiles of the per-iteration corrected AUCs. The same correction is applied to the calibration slope (corrected: 1.0289). As the bootstrap evaluates the raw predictions, a slope close to 1 indicates that predictions do not become too extreme on resampled data. The small optimism therefore shows little evidence of overfitting, within the limits of internal validation on a single dataset.

````python
# Load bootstrap results
boot_xgb = pd.read_csv("bootstrap_results.csv")

# Optimism correction — AUC
opt_xgb_auc      = (boot_xgb["auc_boot_train"] - boot_xgb["auc_boot_test"]).mean()
auc_corrected     = auc_apparent - opt_xgb_auc
corr_per_boot_xgb = auc_apparent - (boot_xgb["auc_boot_train"] - boot_xgb["auc_boot_test"])
ci_xgb            = np.percentile(corr_per_boot_xgb, [2.5, 97.5])

# Optimism correction — calibration slope
opt_xgb_slope  = (boot_xgb["slope_boot_train"] - boot_xgb["slope_boot_test"]).mean()
slope_corrected = slope_apparent - opt_xgb_slope

print(f"=== Optimism-corrected internal validation (B={len(boot_xgb)}) ===")
print(f"Apparent AUC:                          {auc_apparent:.4f}")
print(f"Mean optimism (AUC):                   {opt_xgb_auc:.4f}")
print(f"Optimism-corrected AUC:                {auc_corrected:.4f}")
print(f"Bootstrap-test 95% CI:                 [{ci_xgb[0]:.4f}, {ci_xgb[1]:.4f}]")
print(f"Apparent calibration slope:            {slope_apparent:.4f}")
print(f"Optimism-corrected calibration slope:  {slope_corrected:.4f}")
````

## 4. SHAP interpretability

SHAP (SHapley Additive exPlanations) values are computed with TreeSHAP (path-dependent variant) to assess the contribution of each predictor to individual predictions. Rather than reporting a single global importance score per predictor, SHAP assigns each predictor a value for each observation that reflects its directional effect on the model output: positive values push the prediction towards documented breast cancer history, negative values away from it. Values are in log-odds units and refer to the uncalibrated model; as recalibration preserves the order of predictions, it does not change rankings or directions.

To keep runtime reasonable while preserving representativeness, SHAP values are computed on a random sample of 50,000 rows drawn with replacement in proportion to the BCSC `count` column, so that the sample reflects the distribution of screening exams rather than of the ~429,000 unique covariate patterns. The summary plot shows the predictors ranked by mean absolute SHAP value, with point colour indicating the predictor value (red = high code, blue = low code) and horizontal position the direction and magnitude of the effect. Grey points indicate missing values; the two nominal predictors (race/ethnicity, menopausal status) are also shown in grey, as their codes have no natural order.

````python
# Tested with: xgboost==3.2.0, shap==0.51.0, Python 3.11.7
# Note: SHAP <0.51.0 is incompatible with XGBoost 3.x due to bracket
# notation used for base_score in saved model files.

# Sets sample size and column names to be human-readable for the figure's plot axes.
SAMPLE_SIZE = 50_000
LABELS = {
    "age_group_5_years": "Age (5-yr group)",
    "race_eth": "Race/ethnicity",
    "first_degree_hx": "First-degree family history",
    "age_menarche": "Age at menarche",
    "age_first_birth": "Age at first birth",
    "BIRADS_breast_density": "BI-RADS breast density",
    "current_hrt": "Current HRT use",
    "menopaus": "Menopausal status",
    "bmi_group": "BMI group",
    "biophx": "Prior breast biopsy",
}

# Draws 50,000 rows at random, weighted by `count`, so the sample is representative of the population of screening exams.
rng = np.random.default_rng(42)
probs = df["count"].values / df["count"].values.sum()
sample_idx = rng.choice(len(df), size=SAMPLE_SIZE, replace=True, p=probs)
X_shap = df[FEATURES].iloc[sample_idx].rename(columns=LABELS)

# Restores category dtypes lost during rename
X_shap["Race/ethnicity"] = X_shap["Race/ethnicity"].astype("category")
X_shap["Menopausal status"] = X_shap["Menopausal status"].astype("category")

# Wraps the data in XGBoost's native format and explicitly tells it to handle categorical columns. Without this, SHAP cannot function.
dmat = xgb.DMatrix(X_shap, enable_categorical=True)

# Creates the explainer using the trained model and its booster. 
explainer = shap.TreeExplainer(model.get_booster(), feature_perturbation="tree_path_dependent")
shap_values = explainer.shap_values(dmat, check_additivity=False)

# Designs the SHAP plot
shap.summary_plot(shap_values, X_shap, max_display=10, show=False)
plt.tight_layout()
plt.savefig("fig_shap_summary.png", dpi=200, bbox_inches="tight")
plt.show()
````

## 5. Sensitivity analysis: predictors potentially measured after diagnosis

*Thesis Section 4.8.*

The outcome `breast_cancer_history` records a breast cancer diagnosis **before** the screening exam at which the predictors were collected. Some predictors may therefore partly be a *consequence* of the diagnosis rather than a pre-existing characteristic:

- **Prior breast biopsy (`biophx`)**: a biopsy is required to diagnose breast cancer, so almost every woman with a history of breast cancer will have one.
- **Current HRT use (`current_hrt`)**: hormone therapy is generally not recommended after a breast cancer diagnosis.
- **Menopausal status (`menopaus`)**: some treatments (chemotherapy, endocrine therapy, oophorectomy) can induce menopause; the dataset distinguishes surgical menopause.

To quantify how much the classification relies on these predictors, all three models from Sections 2 and 6 (**XGBoost, Logistic Regression and Random Forest**) are refitted on three nested feature sets:

| Set | Predictors |
|---|---|
| A: full | all 10 predictors (the main models) |
| B: no biopsy | A minus `biophx` |
| C: no biopsy/HRT/menopause | A minus `biophx`, `current_hrt`, `menopaus` |

Each model keeps the settings, class weighting and encoding it has in Sections 2 and 6.

**Bootstrap unit.** The main analyses (Sections 3 and 6) resampled *rows*, i.e. unique covariate patterns, with equal probability regardless of how many screening exams each row represents. Here the bootstrap resamples **individual screening exams**: in each iteration, new row counts are drawn from a multinomial distribution with N = total weighted exams and row probabilities proportional to `count`. This is equivalent to drawing exams with replacement from the expanded dataset, without expanding it. Set A is re-run under this scheme too, so the three sets are compared under the same procedure, and the difference from Sections 3 and 6 shows how much the resampling unit matters.

Note that the BCSC unit of observation is one exam per woman per calendar year, so a woman can contribute up to five exams to the 2013–2017 window. Because the public data contain no woman identifier, repeated exams from the same woman cannot be resampled together; confidence intervals therefore treat exams as independent and are likely somewhat too narrow.

All models and feature sets are fitted on the **same** bootstrap sample in each iteration, so the difference in corrected AUC between sets (ΔAUC) gets a paired percentile confidence interval.

### 5.1 Descriptive check: outcome prevalence by potentially post-diagnostic predictors

Before modelling, the weighted proportion of exams with breast cancer history is tabulated for each level of the three predictors in question, including missing values. A very high prevalence among exams with a prior biopsy, relative to those without, is consistent with biopsy being a downstream marker of the diagnosis.

````python
# Weighted prevalence of breast cancer history per level of each predictor in question
# For one predictor: groups the exams by category (with missing values as their own 'Missing' category),
# then computes the weighted number of exams, how many have documented history, the prevalence within
# the category, and the category's share of all exams with documented history.
def weighted_prevalence(col):
    tmp = df[[col, TARGET, WEIGHT]].copy()
    tmp[col] = tmp[col].astype("object").where(tmp[col].notna(), "Missing")
    g = tmp.groupby(col, dropna=False)
    out = pd.DataFrame({
        "weighted_exams": g[WEIGHT].sum(),
        "with_history":   g.apply(lambda d: d.loc[d[TARGET] == 1, WEIGHT].sum()),
    })
    out["prevalence_%"] = (out["with_history"] / out["weighted_exams"] * 100).round(2)
    out["share_of_all_cases_%"] = (out["with_history"] / out["with_history"].sum() * 100).round(2)
    return out

# Applies the function to the three predictors examined in the sensitivity analysis and saves the tables.
prevalence_tables = {}
for col in ["biophx", "current_hrt", "menopaus"]:
    prevalence_tables[col] = weighted_prevalence(col)
    print(f"\n=== {LABELS[col]} ({col}) ===")
    display(prevalence_tables[col])

pd.concat(prevalence_tables, names=["predictor", "level"]).to_csv("sens_prevalence_by_predictor.csv")
````

### 5.2 Apparent fits for each feature set

Fits XGBoost, Logistic Regression and Random Forest on the full data for each feature set and computes apparent AUC and calibration slope. Logistic Regression and Random Forest use the same one-hot encoded design matrix as in Section 6. The XGBoost models are kept for the SHAP comparison in Section 5.5.

````python
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LogisticRegression as LR
from sklearn.ensemble import RandomForestClassifier

# The three nested predictor sets: A = all ten predictors; B = without prior biopsy;
# C = without prior biopsy, current HRT use and menopausal status.
FEATURE_SETS = {
    "A_full":                FEATURES,
    "B_no_biopsy":           [f for f in FEATURES if f != "biophx"],
    "C_no_biopsy_hrt_menop": [f for f in FEATURES if f not in ("biophx", "current_hrt", "menopaus")],
}
SET_LABELS = {
    "A_full": "A: all predictors",
    "B_no_biopsy": "B: without biopsy",
    "C_no_biopsy_hrt_menop": "C: without biopsy, HRT, menopause",
}
# The three models refitted on every predictor set.
MODELS = ["XGBoost", "Logistic Regression", "Random Forest"]

# One-hot design matrix for Logistic Regression and Random Forest, same encoding as Section 6.1.1
def encoded_matrix(feats):
    cats = [c for c in ["race_eth", "menopaus"] if c in feats]
    nums = [c for c in feats if c not in cats]
    Xd = df[feats].copy()
    for c in cats:
        Xd[c] = Xd[c].astype(str).replace("nan", "Unknown")
    for c in nums:
        Xd[c] = Xd[c].fillna(-1)  # -1 is outside all valid ordinal codes
    transformers = [("pass", "passthrough", nums)]
    if cats:
        transformers.insert(0, ("ohe", OneHotEncoder(handle_unknown="ignore", sparse_output=False), cats))
    return ColumnTransformer(transformers).fit_transform(Xd)

# Pre-computes the encoded matrix once per predictor set, so it is not rebuilt in every bootstrap iteration.
ENC_MATRICES = {name: encoded_matrix(feats) for name, feats in FEATURE_SETS.items()}

# Model constructors: same settings as Sections 2.1 (XGBoost), 6.1 (Logistic Regression) and 6.2 (Random Forest)
def new_model(model_name, seed=42):
    if model_name == "XGBoost":
        return xgb.XGBClassifier(**XGB_PARAMS, scale_pos_weight=scale_pos_weight)
    if model_name == "Logistic Regression":
        return LR(C=1.0, solver="lbfgs", max_iter=1000, class_weight={0: 1, 1: scale_pos_weight})
    if model_name == "Random Forest":
        return RandomForestClassifier(n_estimators=100, max_depth=None,
                                      class_weight={0: 1, 1: scale_pos_weight},
                                      n_jobs=-1, random_state=seed)
    raise ValueError(model_name)

# XGBoost uses the native (categorical) features; the other two models use the encoded matrix
def design(model_name, set_name):
    return df[FEATURE_SETS[set_name]] if model_name == "XGBoost" else ENC_MATRICES[set_name]

# Apparent fits: each model is trained and evaluated on the full data for each predictor set.
# The XGBoost models are stored for the SHAP comparison in Section 5.5.
sens_models, apparent_rows = {}, []
for name in FEATURE_SETS:
    for model_name in MODELS:
        Xd = design(model_name, name)
        m = new_model(model_name).fit(Xd, y, sample_weight=w)
        p = m.predict_proba(Xd)[:, 1]
        slope, _, _ = calibration_slope_intercept(y.values, p, w.values)
        apparent_rows.append({"feature_set": name, "model": model_name,
                              "apparent_auc": weighted_auc(y, p, w), "apparent_slope": slope})
        if model_name == "XGBoost":
            sens_models[name] = m
    print(f"{name}: fitted ({len(FEATURE_SETS[name])} predictors, 3 models)")

apparent_sens = pd.DataFrame(apparent_rows)
apparent_sens
````

### 5.3 Bootstrap internal validation (resampling individual exams)

For each of B iterations, screening exams are resampled with replacement (via multinomial counts), and all three models are refitted for each of the three feature sets on that same resample. Each model is evaluated on the bootstrap sample (`auc_train`) and on the original data (`auc_test`), as in Section 3.

Results are saved per model and feature set to `sens_bootstrap_results.csv`. Re-running the cell resumes where it stopped and computes only what is missing. Because each iteration's resample is fixed by its seed, results from an earlier run (for example XGBoost and Logistic Regression only) are reused, and only the missing models are fitted on the identical resamples.

**Runtime:** Random Forest is the slowest of the three models. Lower `B_SENS` (e.g. 100) for a faster estimate, and report whichever B is used.

````python
# Number of bootstrap iterations, results file, and the key identifying one model fit
# (iteration, predictor set, model).
B_SENS = 200
sens_path = Path("sens_bootstrap_results.csv")
KEY = ["iter", "feature_set", "model"]

# Resume logic: fits already saved are skipped, so an interrupted run continues where it stopped
# and models added later (e.g. Random Forest) reuse the identical resamples.
done_keys = set()
if sens_path.exists():
    prev = pd.read_csv(sens_path).drop_duplicates(subset=KEY, keep="last")
    prev.to_csv(sens_path, index=False)
    done_keys = set(map(tuple, prev[KEY].values))
    print(f"Resuming: {len(done_keys)} of {B_SENS * len(FEATURE_SETS) * len(MODELS)} model fits already completed.")

# Total number of weighted exams and each row's share of them, used for exam-level resampling.
N_exams = int(round(w.sum()))
p_rows = w.values / w.values.sum()
y_arr, w_arr = y.values, w.values

# One bootstrap iteration per b. The seed (5000 + b) fixes the resample, so every model and
# predictor set within an iteration is fitted on exactly the same bootstrap sample (paired comparison).
for b in range(B_SENS):
    todo = [(name, model_name) for name in FEATURE_SETS for model_name in MODELS
            if (b, name, model_name) not in done_keys]
    if not todo:
        continue
    rng = np.random.default_rng(5000 + b)
    # Resamples individual exams: new count per row ~ Multinomial(N, count / N)
    wb = rng.multinomial(N_exams, p_rows).astype(float)
    # Rows not drawn in this resample (count 0) are left out of training.
    keep = wb > 0
    yb, wbk = y_arr[keep], wb[keep]

    iter_rows = []
    # For each remaining (predictor set, model): fit on the bootstrap sample, then evaluate on the
    # bootstrap sample (auc_train, optimistic) and on the original data (auc_test).
    for name, model_name in todo:
        X_full = design(model_name, name)
        X_train = X_full[keep]
        m = new_model(model_name, seed=b).fit(X_train, yb, sample_weight=wbk)
        p_tr = m.predict_proba(X_train)[:, 1]
        p_te = m.predict_proba(X_full)[:, 1]
        s_tr, _, _ = calibration_slope_intercept(yb, p_tr, wbk)
        s_te, _, _ = calibration_slope_intercept(y_arr, p_te, w_arr)
        iter_rows.append({"iter": b, "feature_set": name, "model": model_name,
                          "auc_train": weighted_auc(yb, p_tr, wbk),
                          "auc_test": weighted_auc(y_arr, p_te, w_arr),
                          "slope_train": s_tr, "slope_test": s_te})

    # Saves this iteration's results immediately.
    pd.DataFrame(iter_rows).to_csv(sens_path, mode="a", header=not sens_path.exists(), index=False)
    if (b + 1) % 10 == 0:
        print(f"  iter {b + 1}/{B_SENS} done")

print("Sensitivity bootstrap complete.")
````

### 5.4 Results: optimism-corrected performance per feature set

Corrected AUC and calibration slope are computed as in Section 3.2 (apparent minus mean optimism, with percentile intervals). ΔAUC is the corrected AUC of a reduced set minus that of set A, computed within each bootstrap iteration, so its interval is paired. As in Section 6, Random Forest is compared mainly on discrimination; its calibration slope reflects the extreme probabilities of fully grown trees.

**Note on Random Forest.** For XGBoost and Logistic Regression, the corrected AUC of set A under exam-level resampling matches the row-level bootstrap of Sections 3 and 6. For Random Forest it is markedly higher (≈0.911 vs ≈0.885). With exam-level resampling, nearly every covariate pattern appears in each bootstrap sample, so the memorisation of patterns by fully grown trees is barely penalised and optimism is underestimated (thesis Sections 4.8–4.9).

````python
# Loads the bootstrap results and computes the optimism (train minus test) per fit.
boot_sens = pd.read_csv(sens_path).drop_duplicates(subset=KEY, keep="last")
boot_sens = boot_sens[boot_sens["iter"] < B_SENS]
boot_sens["opt_auc"]   = boot_sens["auc_train"] - boot_sens["auc_test"]
boot_sens["opt_slope"] = boot_sens["slope_train"] - boot_sens["slope_test"]
boot_sens = boot_sens.merge(apparent_sens, on=["feature_set", "model"])
# Corrected AUC per iteration (apparent minus that iteration's optimism), used for the percentile CIs
# and for the paired ΔAUC below.
boot_sens["corr_auc_b"] = boot_sens["apparent_auc"] - boot_sens["opt_auc"]

# Summary per model and predictor set: corrected AUC = apparent AUC minus mean optimism,
# with a 95% percentile interval, and the corrected calibration slope.
summary_rows = []
for model_name in MODELS:
    for name in FEATURE_SETS:
        g = boot_sens[(boot_sens["feature_set"] == name) & (boot_sens["model"] == model_name)]
        if g.empty:
            continue
        app = apparent_sens.query("feature_set == @name and model == @model_name").iloc[0]
        lo, hi = np.percentile(g["corr_auc_b"], [2.5, 97.5])
        summary_rows.append({
            "Model": model_name, "Feature set": SET_LABELS[name],
            "Apparent AUC": app["apparent_auc"],
            "Corrected AUC": app["apparent_auc"] - g["opt_auc"].mean(),
            "95% CI low": lo, "95% CI high": hi,
            "Corrected cal. slope": app["apparent_slope"] - g["opt_slope"].mean(),
            "B": g["iter"].nunique(),
        })
sens_summary = pd.DataFrame(summary_rows)

# Paired difference vs set A, per model
delta_rows = []
wide = boot_sens.pivot_table(index=["iter", "model"], columns="feature_set", values="corr_auc_b").reset_index()
for model_name in MODELS:
    g = wide[wide["model"] == model_name].dropna()
    if g.empty:
        continue
    for name in ["B_no_biopsy", "C_no_biopsy_hrt_menop"]:
        d = g[name] - g["A_full"]
        lo, hi = np.percentile(d, [2.5, 97.5])
        delta_rows.append({"Model": model_name, "Comparison": f"{SET_LABELS[name]} vs A",
                           "ΔAUC (mean)": d.mean(), "95% CI low": lo, "95% CI high": hi})
sens_delta = pd.DataFrame(delta_rows)

# Saves both tables for use in the thesis.
sens_summary.round(4).to_csv("sens_summary.csv", index=False)
sens_delta.round(4).to_csv("sens_delta_auc.csv", index=False)
display(sens_summary.round(4))
display(sens_delta.round(4))
````

### 5.6 Mean SHAP per category, including missing values

Section 5.1 shows how outcome prevalence differs between categories **in the data**. This section shows how the main XGBoost model **uses** those categories: for each level of biopsy history, HRT use and menopausal status, including missing values, it reports the mean signed SHAP value (positive = pushes the prediction towards documented history, negative = away from it) and the mean absolute SHAP value. SHAP dependence plots are not used, as all predictors are categorical with few levels; a per-category summary conveys the same information and reports missing values explicitly.

````python
# Readable names for the category codes (per the BCSC data dictionary). Code 3 of menopaus = surgical menopause.
LEVEL_NAMES = {
    "biophx":      {0: "No", 1: "Yes"},
    "current_hrt": {0: "No", 1: "Yes"},
    "menopaus":    {1: "Pre-/perimenopausal", 2: "Postmenopausal", 3: "Surgical menopause"},
}

# SHAP values of the main model (Section 4) and the raw predictor values of the same 50,000 sampled rows.
shap_df = pd.DataFrame(np.asarray(shap_values), columns=FEATURES)
raw_vals = df[FEATURES].iloc[sample_idx].reset_index(drop=True)

rows = []
# For each predictor and category (including 'Missing'): share of the sample, mean signed SHAP value
# (direction of the effect) and mean absolute SHAP value (size of the effect).
for col, names in LEVEL_NAMES.items():
    level = raw_vals[col].astype("object").map(lambda v: "Missing" if pd.isna(v) else names.get(int(v), str(v)))
    for lv, g in shap_df[col].groupby(level):
        rows.append({"Predictor": LABELS[col], "Category": lv,
                     "Share of sample (%)": round(len(g) / len(shap_df) * 100, 1),
                     "Mean SHAP": round(g.mean(), 3),
                     "Mean |SHAP|": round(g.abs().mean(), 3)})

shap_by_category = pd.DataFrame(rows)
# Sorts categories in a fixed, readable order with 'Missing' last.
order = {"No": 0, "Yes": 1, "Pre-/perimenopausal": 0, "Postmenopausal": 1, "Surgical menopause": 2, "Missing": 9}
shap_by_category["_o"] = shap_by_category["Category"].map(order)
shap_by_category = (shap_by_category.sort_values(["Predictor", "_o"]).drop(columns="_o")
                    .reset_index(drop=True))
shap_by_category.to_csv("sens_shap_by_category.csv", index=False)
print("SHAP values are in log-odds units of the uncalibrated model; positive = towards documented history.")
display(shap_by_category)
````

## 6. Comparison with Logistic Regression and Random Forest

*Thesis Section 4.9.*

To place the XGBoost model's performance in context, two comparator models are trained on the same data using identical weighting, class-imbalance handling and bootstrap validation: a Logistic Regression baseline and a Random Forest ensemble. Several other candidate models were considered but excluded. Naïve Bayes assumes conditional independence between predictors, an assumption clearly violated by the correlations among age, BMI, menopausal status and HRT use in this dataset. K-Nearest Neighbours is computationally intractable at 428,808 rows and cannot handle sample weights meaningfully. Support Vector Machines do not produce calibrated probabilities natively and do not scale to this dataset size without approximations that would undermine a fair comparison.

### 6.1.1 Logistic Regression - Apparent AUC

Logistic Regression serves as the linear baseline. Because it cannot use pandas `category` dtype natively, the two nominal predictors (`race_eth` and `menopaus`) are one-hot encoded via a `ColumnTransformer`, and remaining missing values are filled with a sentinel value of −1, which falls outside all valid ordinal codes. Class imbalance is addressed using `class_weight={0: 1, 1: scale_pos_weight}`, mirroring the `scale_pos_weight` parameter used in XGBoost. The model is fitted with L2 regularisation (C = 1.0) using the lbfgs solver. Apparent weighted AUC on the full training data is **0.8685**, with a calibration slope of 1.0001 and intercept of −2.2297 — the near-unity slope indicates that Logistic Regression's predicted probabilities are already well-spread, with only an intercept offset introduced by class weighting. Logistic recalibration is then applied using the same Cox calibration approach from Section 2, correcting the intercept shift.

````python
from sklearn.linear_model import LogisticRegression as LR
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

# Encodes categoricals - XGBoost handled NaN natively, LR needs numeric input
cat_cols = ["race_eth", "menopaus"]
num_cols = [c for c in FEATURES if c not in cat_cols]

preprocessor = ColumnTransformer([
    ("ohe", OneHotEncoder(handle_unknown="ignore", sparse_output=False), cat_cols),
    ("pass", "passthrough", num_cols),
])

X_lr = X.copy()
for c in cat_cols:
    X_lr[c] = X_lr[c].astype(str).fillna("Unknown").replace("nan", "Unknown")
for c in num_cols:
    X_lr[c] = X_lr[c].fillna(-1)  # -1 is outside all valid ordinal codes

X_lr_enc = preprocessor.fit_transform(X_lr)

# Training - class_weight mirrors XGBoost's scale_pos_weight
lr_model = LR(C=1.0, solver="lbfgs", max_iter=1000,
              class_weight={0: 1, 1: scale_pos_weight})
lr_model.fit(X_lr_enc, y, sample_weight=w)

# Apparent performance
p_lr_apparent = lr_model.predict_proba(X_lr_enc)[:, 1]
auc_lr_apparent = weighted_auc(y, p_lr_apparent, w)
slope_lr, intercept_lr, logit_lr = calibration_slope_intercept(y, p_lr_apparent, w)
p_lr_recal = 1 / (1 + np.exp(-(slope_lr * logit_lr + intercept_lr)))

print(f"LR apparent AUC:          {auc_lr_apparent:.4f}")
print(f"LR calibration slope:     {slope_lr:.4f}")
print(f"LR calibration intercept: {intercept_lr:.4f}")
````

### 6.1.2 Logistic Regression - Bootstrap Validation and optimism correction

Bootstrap internal validation (B = 200) follows the same Harrell optimism-correction procedure used for XGBoost in Section 3. In each iteration, a new Logistic Regression model is trained on a bootstrap resample of the preprocessed encoded matrix (X_lr_enc) and evaluated on both the bootstrap sample and the original full dataset. The difference in AUC between the two evaluations estimates the optimism for that iteration. Mean optimism across all 200 iterations is subtracted from the apparent AUC to yield the optimism-corrected AUC of 0.8684 (95% CI: 0.8661–0.8706), and the same correction is applied to the calibration slope, yielding 0.9995. The negligible optimism (apparent AUC 0.8685 vs corrected 0.8684) and near-unity calibration slope confirm that Logistic Regression is not overfitting to the training data: its in-sample and out-of-sample performance are essentially identical.

````python
B = 200
lr_boot_path = Path("bootstrap_lr_results.csv")
done_lr = set()
if lr_boot_path.exists():
    done_lr = set(pd.read_csv(lr_boot_path)["iter"].tolist())
    print(f"Resuming LR bootstrap: {len(done_lr)} iterations already completed.")

lr_rows = []
n = len(df)

for b in range(B):
    if b in done_lr:
        continue
    rng = np.random.default_rng(b)
    idx = rng.integers(0, n, size=n)

    X_b, y_b, w_b = X_lr_enc[idx], y.values[idx], w.values[idx]

    m = LR(C=1.0, solver="lbfgs", max_iter=1000,
           class_weight={0: 1, 1: scale_pos_weight})
    m.fit(X_b, y_b, sample_weight=w_b)

    p_train = m.predict_proba(X_b)[:, 1]
    p_test  = m.predict_proba(X_lr_enc)[:, 1]

    auc_train = weighted_auc(y_b, p_train, w_b)
    auc_test  = weighted_auc(y, p_test, w)
    sl_train, _, _ = calibration_slope_intercept(y_b, p_train, w_b)
    sl_test,  _, _ = calibration_slope_intercept(y,  p_test,  w)

    lr_rows.append({"iter": b, "auc_boot_train": auc_train, "auc_boot_test": auc_test,
                    "slope_boot_train": sl_train, "slope_boot_test": sl_test})

    if (b + 1) % 10 == 0:
        print(f"  LR bootstrap {b+1}/{B}")
        tmp = pd.DataFrame(lr_rows)
        if lr_boot_path.exists():
            tmp = pd.concat([pd.read_csv(lr_boot_path), tmp], ignore_index=True)
        tmp.to_csv(lr_boot_path, index=False)
        lr_rows = []

if lr_rows:
    tmp = pd.DataFrame(lr_rows)
    if lr_boot_path.exists():
        tmp = pd.concat([pd.read_csv(lr_boot_path), tmp], ignore_index=True)
    tmp.to_csv(lr_boot_path, index=False)

# Optimism correction
boot_lr = pd.read_csv(lr_boot_path)
opt_lr_auc    = (boot_lr["auc_boot_train"] - boot_lr["auc_boot_test"]).mean()
auc_lr_corr   = auc_lr_apparent - opt_lr_auc
corr_per_boot_lr = auc_lr_apparent - (boot_lr["auc_boot_train"] - boot_lr["auc_boot_test"])
ci_lr         = np.percentile(corr_per_boot_lr, [2.5, 97.5])
opt_lr_slope  = (boot_lr["slope_boot_train"] - boot_lr["slope_boot_test"]).mean()
slope_lr_corr = slope_lr - opt_lr_slope

print(f"\nLR optimism-corrected AUC: {auc_lr_corr:.4f}  (95% CI: {ci_lr[0]:.4f}–{ci_lr[1]:.4f})")
print(f"LR corrected cal. slope:   {slope_lr_corr:.4f}")
````

### 6.2.1 Random Forest - Apparent AUC

A Random Forest classifier is trained using 100 fully grown trees (max_depth=None) on the same one-hot encoded feature matrix (X_lr_enc) used for Logistic Regression. Class imbalance is handled identically, via class_weight={0: 1, 1: scale_pos_weight}. Apparent weighted AUC on the full training data is 0.9305, markedly higher than both Logistic Regression (0.8685) and XGBoost (0.8968). This is expected: fully grown trees memorise their training samples, and apparent performance on training data reflects this overfitting rather than true generalisability. The apparent calibration slope is 1.0876 with intercept −2.3328; logistic recalibration is applied to correct the intercept offset before performing the final comparison.

````python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=100,
    max_depth=None,
    class_weight={0: 1, 1: scale_pos_weight},
    n_jobs=-1,
    random_state=42,
)
rf_model.fit(X_lr_enc, y, sample_weight=w)

# Apparent performance
p_rf_apparent = rf_model.predict_proba(X_lr_enc)[:, 1]
auc_rf_apparent = weighted_auc(y, p_rf_apparent, w)
slope_rf, intercept_rf, logit_rf = calibration_slope_intercept(y, p_rf_apparent, w)
p_rf_recal = 1 / (1 + np.exp(-(slope_rf * logit_rf + intercept_rf)))

print(f"RF apparent AUC:          {auc_rf_apparent:.4f}")
print(f"RF calibration slope:     {slope_rf:.4f}")
print(f"RF calibration intercept: {intercept_rf:.4f}")
````

### 6.2.2 Random Forest - Bootstrap Validation and optimism correction

Bootstrap internal validation (B = 200) reveals the extent of Random Forest's overfitting. The optimism-corrected AUC drops to 0.8849 (95% CI: 0.8826–0.8871), a reduction of 0.0456 from the apparent value and by far the largest optimism of the three models. The corrected calibration slope is 0.5278, indicating that its predictions are expected to be far too extreme on new data: probabilities are pushed too close to 0 and 1. This reflects the memorisation of covariate patterns by fully grown trees. Its discrimination is also sensitive to the validation design: under the exam-level bootstrap of Section 5, its corrected AUC is higher (≈0.911), because nearly every pattern then appears in each bootstrap sample and memorisation is barely penalised.

````python
# Bootstrap validation
rf_boot_path = Path("bootstrap_rf_results.csv")
done_rf = set()
if rf_boot_path.exists():
    done_rf = set(pd.read_csv(rf_boot_path)["iter"].tolist())
    print(f"Resuming RF bootstrap: {len(done_rf)} iterations already completed.")

rf_rows = []

for b in range(B):
    if b in done_rf:
        continue
    rng = np.random.default_rng(b)
    idx = rng.integers(0, n, size=n)

    X_b, y_b, w_b = X_lr_enc[idx], y.values[idx], w.values[idx]

    m = RandomForestClassifier(
        n_estimators=100,
        class_weight={0: 1, 1: scale_pos_weight},
        n_jobs=-1,
        random_state=b,
    )
    m.fit(X_b, y_b, sample_weight=w_b)

    p_train = m.predict_proba(X_b)[:, 1]
    p_test  = m.predict_proba(X_lr_enc)[:, 1]

    auc_train = weighted_auc(y_b, p_train, w_b)
    auc_test  = weighted_auc(y, p_test, w)
    sl_train, _, _ = calibration_slope_intercept(y_b, p_train, w_b)
    sl_test,  _, _ = calibration_slope_intercept(y,  p_test,  w)

    rf_rows.append({"iter": b, "auc_boot_train": auc_train, "auc_boot_test": auc_test,
                    "slope_boot_train": sl_train, "slope_boot_test": sl_test})

    if (b + 1) % 10 == 0:
        print(f"  RF bootstrap {b+1}/{B}")
        tmp = pd.DataFrame(rf_rows)
        if rf_boot_path.exists():
            tmp = pd.concat([pd.read_csv(rf_boot_path), tmp], ignore_index=True)
        tmp.to_csv(rf_boot_path, index=False)
        rf_rows = []

if rf_rows:
    tmp = pd.DataFrame(rf_rows)
    if rf_boot_path.exists():
        tmp = pd.concat([pd.read_csv(rf_boot_path), tmp], ignore_index=True)
    tmp.to_csv(rf_boot_path, index=False)

# Optimism correction
boot_rf = pd.read_csv(rf_boot_path)
opt_rf_auc    = (boot_rf["auc_boot_train"] - boot_rf["auc_boot_test"]).mean()
auc_rf_corr   = auc_rf_apparent - opt_rf_auc
corr_per_boot_rf = auc_rf_apparent - (boot_rf["auc_boot_train"] - boot_rf["auc_boot_test"])
ci_rf         = np.percentile(corr_per_boot_rf, [2.5, 97.5])
opt_rf_slope  = (boot_rf["slope_boot_train"] - boot_rf["slope_boot_test"]).mean()
slope_rf_corr = slope_rf - opt_rf_slope

print(f"\nRF optimism-corrected AUC: {auc_rf_corr:.4f}  (95% CI: {ci_rf[0]:.4f}–{ci_rf[1]:.4f})")
print(f"RF corrected cal. slope:   {slope_rf_corr:.4f}")
````
